```python
import numpy as np
from matplotlib import pyplot as plt

#nous importons l'image , et nous reduisons ses dimentions à 2 
image = plt.imread('bacteria.png')
image = image[:,:,0]
plt.imshow(image)
image.shape
```

```python
#on créé une copie de notre image , puis on applatis cette copie sous forme d'un histogramme 
image_b= np.copy(image)
plt.hist(image_b.ravel() , bins = 255 ) 
plt.show()
#en observant l'histogramme , et en considérant que la partie horizontale represente l'intensité des couleurs on voit que les parties les plus sombres sont proches de l'index 0 et les parties claires sont de plus en plus eloignés de cet index ... 
```

```python
#filtrans les parties sombres par du booleen indexing 
image= image < 0.6
image
plt.imshow(image)
```

```python
#debarassons nous des artéfacts 
from scipy import ndimage
open_x = ndimage.binary_opening(image)
plt.imshow(image)
```

```python
#utilisons la fonction label de ndimage pour voir le nombre de groupe sur l'image (nlabel), et ler representation (label_image)
label_image, n_labels = ndimage.label(open_x)
print(n_labels)
```

```python
plt.imshow(label_image)
```

```python
#avec la fonction sum du module ndimage essayons de voir le nombre de pixel a l'interieru de chaque groupe et seeayons de les representer 
sizes = ndimage.sum(open_x, label_image , range(n_labels))
```

```python
sizes
```

```python
plt.scatter(range(n_labels), sizes, c= 'blue')
```
