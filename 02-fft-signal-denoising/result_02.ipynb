```python
import numpy as np
from matplotlib import pyplot as plt
from scipy import fftpack

# Générons un signal propre
x = np.linspace(0, 30 , 1000)
y_propre = 3*np.sin(x) + 2*np.sin(5*x) + np.sin(10*x)

# Génération d'un bruit blanc gaussien
# (moyenne = 0, écart-type/intensité du bruit = 1.5, taille identique à x)
bruit = np.random.normal(0, 1.5, size=x.shape)

# Signal bruité (on ajoute le bruit au signal d'origine)
y = y_propre + bruit

# Affichage du signal bruité
plt.plot(x, y)
plt.title("Signal bruité")
plt.xlabel("Temps")
plt.ylabel("Amplitude")
plt.show()
```

```python
#utilisons la transformation de fourrier pour produire le spectre de notre signel
fourier = fftpack.fft(y)
power = np.abs(fourier)
frequences = fftpack.fftfreq(y.size)
plt.plot(np.abs(frequences), power)
```

```python
#utilisons du booléan indexing pour eliminé les valeurs inferieurs a 400
fourier[power <400] =0
plt.plot(np.abs(frequences), np.abs(fourier))
```

```python
#utilisé la transformation de fourier inverse sur le spectre obtenu pour avoir ainsi un signal clean
#la transformation de fourier inverse se fai avec la fonction ifft
filtered_signal=fftpack.ifft(fourier)
plt.plot(filtered_signal)
```
